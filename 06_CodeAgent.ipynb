```python
from smolagents import CodeAgent, LiteLLMModel, tool
```

```python
@tool
def get_temperature(city: str) -> float:
    """
    Return the temperature in Celsius for a given city.

    Args:
        city: The city to get the temperature for.
    """
    temperatures = {
        "istanbul": 18.0,
        "ankara": 14.0,
        "izmir": 24.0,
    }

    return temperatures.get(city.lower(), 20.0)
```

```python
@tool
def celsius_to_fahrenheit(celsius: float) -> float:
    """
    Convert a Celsius temperature to Fahrenheit.

    Args:
        celsius: The temperature in Celsius.
    """
    return celsius * 9 / 5 + 32
```

```python
model = LiteLLMModel(
    model_id="ollama_chat/qwen3.5:9b",
    api_base="http://localhost:11434",
)
```

```python
@tool
def get_temperature(city: str) -> float:
    """
    Return the temperature in Celsius for a given city.

    Args:
        city: The city to get the temperature for.
    """
    temperatures = {
        "istanbul": 18.0,
        "ankara": 14.0,
        "izmir": 24.0,
    }

    return temperatures.get(city.lower(), 20.0)
```

```python
agent = CodeAgent(
    tools=[
        get_temperature,
        celsius_to_fahrenheit,
    ],
    model=model,
    max_steps=5,
)
```

```python
result = agent.run(
    "İstanbul'un sıcaklığını bul ve Fahrenheit'a çevir."
)

print(result)
```

```python
from smolagents import ActionStep

for step in agent.memory.steps:
    if isinstance(step, ActionStep):
        print(f"Step {step.step_number}")
        print("Tool calls:", step.tool_calls)
        print("Observation:", step.observations)
        print()
```

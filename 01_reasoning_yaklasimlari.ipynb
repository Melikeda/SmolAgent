```python
task = """
Bir ürünün fiyatı 1.200 TL'dir.
Önce yüzde 10 indirim uygulanıyor.
Daha sonra 80 TL kargo ekleniyor.
Müşteri toplam kaç TL öder?
"""

prompts = {
    "Doğrudan cevap": f"""
{task}

Yalnızca final sonucu tek cümleyle yaz.
""",

    "Adım adım reasoning": f"""
{task}

Problemi küçük adımlara ayır.
Her işlemin sonucunu kısa şekilde göster.
Final cevabı ayrı bir satırda yaz.
""",

    "Plan and Execute": f"""
{task}

Önce kısa bir çözüm planı oluştur.
Daha sonra planı sırayla uygula.
Çıktıyı Plan, Uygulama ve Sonuç başlıklarıyla yaz.
""",

    "Reflection": f"""
{task}

Problemi çöz.
Daha sonra işlemleri ikinci kez kontrol et.
Yalnızca final sonucu ve kısa kontrol özetini yaz.
"""
}
```

```python
for approach, prompt in prompts.items():
    response = chat(
        model="qwen3.5:9b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        think=True,
        stream=False
    )

    print("\n" + "=" * 60)
    print(approach.upper())
    print("=" * 60)
    print(response.message.content)
```

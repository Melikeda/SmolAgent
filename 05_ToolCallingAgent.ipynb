```python
from smolagents import tool
```

```python
@tool
def multiply(a: int, b: int) -> int:
    """
    Multiply two integers.

    Args:
        a: The first integer.
        b: The second integer.
    """
    return a * b
```

```python
print(multiply.name)
```

```python
print(multiply.description)
```

```python
print(multiply.inputs)
```

```python
print(multiply.output_type)
```

```python
agent = ToolCallingAgent(
    tools=[multiply],
    model=model,
)
```

```python
result = agent.run(
    "47 ile 23'ü çarp."
)

print(result)
```

```python
@tool
def get_temperature(city: str) -> float:
    """
    Return the temperature in Celsius for a given city.

    Args:
        city: The city to get the temperature for.
    """
    temperatures = {
        "istanbul": 18.0,
        "ankara": 14.0,
        "izmir": 24.0,
    }

    return temperatures.get(city.lower(), 20.0)
```

```python
@tool
def celsius_to_fahrenheit(celsius: float) -> float:
    """
    Convert a Celsius temperature to Fahrenheit.

    Args:
        celsius: The temperature in Celsius.
    """
    return celsius * 9 / 5 + 32
```

```python
agent = ToolCallingAgent(
    tools=[
        multiply,
        get_temperature,
        celsius_to_fahrenheit,
    ],
    model=model,
)
```

```python
result = agent.run(
    "27 ile 14'ü çarp."
)

print(result)
```

```python
result = agent.run(
    "İstanbul'un sıcaklığını getir."
)

print(result)
```

```python
result = agent.run(
    "20 Celsius kaç Fahrenheit?"
)

print(result)
```

```python
result = agent.run(
    "İstanbul'un sıcaklığını bul ve Fahrenheit'a çevir."
)

print(result)
```

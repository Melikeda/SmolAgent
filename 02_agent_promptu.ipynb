```python
weak_prompt = """
Sen yardımcı bir müşteri destek asistanısın.
"""
```

```python
structured_prompt = """
ROL

Sen bir müşteri destek agentısın.

HEDEF

Kullanıcının sipariş sorununu anlamak
ve bir sonraki doğru adımı belirlemek.

BAĞLAM

Sipariş durumunu kontrol etmek için
sipariş numarası gereklidir.

KURALLAR

Eksik bilgiyi tahmin etme.
Sipariş numarası yoksa kullanıcıdan iste.
Aynı anda yalnızca bir soru sor.

ÇALIŞMA BİÇİMİ

Önce gerekli bilgilerin mevcut olup
olmadığını kontrol et.

DURMA KOŞULU

Bir sonraki adım belli olduğunda cevap ver.

ÇIKTI FORMATI

Türkçe yaz.
Tek ve kısa bir soru veya cevap üret.
"""
```

```python
user_message = "Siparişim hâlâ gelmedi."
```

```python
from ollama import chat

prompts = {
    "Belirsiz prompt": weak_prompt,
    "Yapılandırılmış prompt": structured_prompt
}

for name, system_prompt in prompts.items():
    response = chat(
        model="qwen3.5:9b",
        messages=[
            {
                "role": "system",
                "content": system_prompt
            },
            {
                "role": "user",
                "content": user_message
            }
        ],
        think=True,
        stream=False
    )

    print("\n" + "=" * 60)
    print(name.upper())
    print("=" * 60)
    print(response.message.content)
```
